Experiment-05

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1,2,2)

for i, c in enumerate(['b','g','r']):
    hist = cv2.calcHist([img],[i],None,[256],[0,256])
    plt.plot(hist, color=c)

plt.title("Color Histogram")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0], 0)

_, img = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

kernel = np.ones((5,5), np.uint8)
erode = cv2.erode(img, kernel, iterations=1)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(erode, cmap='gray')
plt.title("Erosion")
plt.show()
```

Experiment-07

```python
import cv2
from google.colab import files
from IPython.display import Video, display

f = files.upload()
name = list(f.keys())[0]

cap = cv2.VideoCapture(name)
fps = cap.get(cv2.CAP_PROP_FPS)
w = int(cap.get(3))
h = int(cap.get(4))

for speed, output in [(2,"slow.mp4"),(0.5,"fast.mp4")]:
    cap.set(cv2.CAP_PROP_POS_FRAMES,0)

    out = cv2.VideoWriter(output,
        cv2.VideoWriter_fourcc(*'mp4v'),
        fps/speed,(w,h))

    while True:
        ret, frame = cap.read()
        if not ret:
            break
        out.write(frame)

    out.release()

cap.release()

print("Original Video")
display(Video(name, embed=True))

print("Slow Motion")
display(Video("slow.mp4", embed=True))

print("Fast Motion")
display(Video("fast.mp4", embed=True))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0], 0)

_, img = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

kernel = np.ones((5,5), np.uint8)
dilate = cv2.dilate(img, kernel, iterations=1)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(dilate, cmap='gray')
plt.title("Dilation")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])

big = cv2.resize(img,None,fx=2,fy=2)
small = cv2.resize(img,None,fx=0.5,fy=0.5)

plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(big,cv2.COLOR_BGR2RGB))
plt.title("Big")

plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(small,cv2.COLOR_BGR2RGB))
plt.title("Small")

plt.show()
```
