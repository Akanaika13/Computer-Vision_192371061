Experiment-31

```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]

img = cv2.imread(filename)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

ret, result = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

print("Original Image:")
cv2_imshow(img)

print("Segmented Image:")
cv2_imshow(result)
```

Experiment-32

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def create_image(width, height):
    img = np.ones((height, width, 3), dtype=np.uint8) * 255

    w = width // 10
    h = height // 10

    img[0:h, 0:w] = (0, 0, 0)
    img[0:h, width-w:width] = (255, 0, 0)
    img[height-h:height, 0:w] = (0, 255, 0)
    img[height-h:height, width-w:width] = (0, 0, 255)

    cv2_imshow(img)

width = int(input("Enter width: "))
height = int(input("Enter height: "))

create_image(width, height)
```

Experiment-33

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def create_rectangle(width, height):
    img = np.ones((height, width, 3), dtype=np.uint8) * 255

    cv2.rectangle(img, (50, 50), (200, 150), (0, 0, 255), 3)

    cv2_imshow(img)

w = int(input("Enter width: "))
h = int(input("Enter height: "))

create_rectangle(w, h)
```

Experiment-34

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def create_circle(width, height):
    img = np.ones((height, width, 3), dtype=np.uint8) * 255

    center = (width // 2, height // 2)
    radius = min(width, height) // 4

    cv2.circle(img, center, radius, (0, 0, 255), 3)

    cv2_imshow(img)

w = int(input("Enter width: "))
h = int(input("Enter height: "))

create_circle(w, h)
```

Experiment-35

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

def add_text():
    uploaded = files.upload()
    filename = list(uploaded.keys())[0]

    img = cv2.imread(filename)

    text = input("Enter text: ")

    cv2.putText(img, text, (50, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                1, (0, 0, 255), 2)

    cv2_imshow(img)

add_text()
```
