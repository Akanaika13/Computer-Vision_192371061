Experiment-16

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0],0)

sobelx = cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
sobely = cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)

sobel = cv2.magnitude(sobelx,sobely)

plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(sobel,cmap='gray')
plt.title("Sobel Edge")
plt.show()
```

Experiment-17

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])

watermark = img.copy()

cv2.putText(watermark,"MY WATERMARK",(30,50),
            cv2.FONT_HERSHEY_SIMPLEX,1,
            (0,0,255),2)

result = cv2.addWeighted(img,0.7,watermark,0.3,0)

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(result,cv2.COLOR_BGR2RGB))
plt.title("Watermarked")
plt.show()
```

Experiment-18

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])

roi = img[50:200,50:200]

cropped = roi.copy()

result = img.copy()
result[0:150,0:150] = cropped

plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(roi,cv2.COLOR_BGR2RGB))
plt.title("Cropped ROI")

plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(result,cv2.COLOR_BGR2RGB))
plt.title("Copied and Pasted")
plt.show()
```

Experiment-19

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0],0)

_, binary = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

kernel = np.ones((5,5),np.uint8)

erosion = cv2.morphologyEx(binary,cv2.MORPH_ERODE,kernel)

plt.subplot(1,2,1)
plt.imshow(binary,cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(erosion,cmap='gray')
plt.title("Erosion")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0],0)

_, binary = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

kernel = np.ones((5,5),np.uint8)

dilation = cv2.morphologyEx(binary,cv2.MORPH_DILATE,kernel)

plt.subplot(1,2,1)
plt.imshow(binary,cmap='gray')
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(dilation,cmap='gray')
plt.title("Dilation")
plt.show()
```
