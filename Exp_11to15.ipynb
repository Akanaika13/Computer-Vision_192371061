Experiment-13

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])

rows, cols = img.shape[:2]

p1 = np.float32([[50,50],[200,50],[50,200]])
p2 = np.float32([[10,100],[200,50],[100,250]])

M = cv2.getAffineTransform(p1,p2)

result = cv2.warpAffine(img,M,(cols,rows))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(result,cv2.COLOR_BGR2RGB))
plt.title("Affine Transformation")
plt.show()
```

Experiment-14

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])

h, w = img.shape[:2]

p1 = np.float32([[0,0],[w,0],[0,h],[w,h]])
p2 = np.float32([[50,50],[w-50,0],[0,h-50],[w-50,h-50]])

M = cv2.getPerspectiveTransform(p1,p2)

result = cv2.warpPerspective(img,M,(w,h))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(result,cv2.COLOR_BGR2RGB))
plt.title("Perspective")
plt.show()
```

Experiment-15

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

f = files.upload()
img = cv2.imread(list(f.keys())[0])

gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)

corners = cv2.cornerHarris(gray,2,3,0.04)

img[corners > 0.01 * corners.max()] = [0,0,255]

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(cv2.imread(list(f.keys())[0]),cv2.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Harris Corners")
plt.show()
```
