```python
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

uploaded = files.upload()

filename = list(uploaded.keys())[0]

img = cv2.imread(filename, 0)

if img is None:
    print("Image not found")
else:
    kernel = np.ones((5,5), np.uint8)

    output = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)

    print("Original Image")
    cv2_imshow(img)

    print("Opening Output")
    cv2_imshow(output)
```

Experiment-25

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

print("Upload Original Image")
a = files.upload()
img = cv2.imread(list(a.keys())[0])

print("Upload Watch Template")
b = files.upload()
temp = cv2.imread(list(b.keys())[0])

if img is None or temp is None:
    print("Invalid image")
elif temp.shape[0] > img.shape[0] or temp.shape[1] > img.shape[1]:
    print("Template must be smaller than original image")
else:
    result = cv2.matchTemplate(img, temp, cv2.TM_CCOEFF_NORMED)
    _, score, _, loc = cv2.minMaxLoc(result)

    h, w = temp.shape[:2]
    cv2.rectangle(img, loc, (loc[0]+w, loc[1]+h), (0,255,0), 3)

    print("Original Image")
    cv2_imshow(cv2.imread(list(a.keys())[0]))

    print("Watch Recognition Output")
    cv2_imshow(img)
    print("Match Score:", score)
```
